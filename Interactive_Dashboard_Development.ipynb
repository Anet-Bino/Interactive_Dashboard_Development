```python
df = pd.read_csv(
    '/content/superstore/Sample - Superstore.csv',
    encoding='latin1'
)

df.head()
```

```python
# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

print("\nDataset after cleaning:", df.shape)
```

```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_orders = df['Order ID'].nunique()
average_sales = df['Sales'].mean()

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Average Sales:", round(average_sales, 2))
```

```python
# Convert Order Date into datetime format
df['Order Date'] = pd.to_datetime(df['Order Date'])

# Group sales by month
monthly_sales = df.groupby(
    df['Order Date'].dt.to_period('M')
)['Sales'].sum().reset_index()

monthly_sales['Order Date'] = monthly_sales['Order Date'].astype(str)

# Create interactive line chart
fig = px.line(
    monthly_sales,
    x='Order Date',
    y='Sales',
    title='Monthly Sales Trend',
    markers=True,
    labels={
        'Order Date': 'Month',
        'Sales': 'Total Sales'
    }
)

fig.update_layout(
    template='plotly_white',
    xaxis_title='Month',
    yaxis_title='Sales',
    hovermode='x unified'
)

fig.show()
```

```python
# Calculate sales by category
category_sales = df.groupby('Category')['Sales'].sum().reset_index()

# Create interactive bar chart
fig = px.bar(
    category_sales,
    x='Category',
    y='Sales',
    color='Category',
    title='Sales by Product Category',
    text_auto='.2s'
)

fig.update_layout(
    template='plotly_white',
    xaxis_title='Product Category',
    yaxis_title='Total Sales'
)

fig.show()
```

```python
# Calculate sales by region
region_sales = df.groupby('Region')['Sales'].sum().reset_index()

# Create interactive bar chart
fig = px.bar(
    region_sales,
    x='Region',
    y='Sales',
    color='Region',
    title='Sales Performance by Region',
    text_auto='.2s'
)

fig.update_layout(
    template='plotly_white',
    xaxis_title='Region',
    yaxis_title='Total Sales'
)

fig.show()
```

```python
# Calculate sales by product
top_products = (
    df.groupby('Product Name')['Sales']
    .sum()
    .reset_index()
    .sort_values(by='Sales', ascending=False)
    .head(10)
)

# Create interactive bar chart
fig = px.bar(
    top_products,
    x='Sales',
    y='Product Name',
    orientation='h',
    title='Top 10 Products by Sales',
    color='Sales',
    text_auto='.2s'
)

fig.update_layout(
    template='plotly_white',
    yaxis={'categoryorder': 'total ascending'},
    xaxis_title='Total Sales',
    yaxis_title='Product Name'
)

fig.show()
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Create dropdown filters
region_filter = widgets.Dropdown(
    options=['All'] + sorted(df['Region'].unique().tolist()),
    value='All',
    description='Region:'
)

category_filter = widgets.Dropdown(
    options=['All'] + sorted(df['Category'].unique().tolist()),
    value='All',
    description='Category:'
)

output = widgets.Output()

# Dashboard update function
def update_dashboard(change=None):

    with output:
        clear_output(wait=True)

        filtered_df = df.copy()

        if region_filter.value != 'All':
            filtered_df = filtered_df[
                filtered_df['Region'] == region_filter.value
            ]

        if category_filter.value != 'All':
            filtered_df = filtered_df[
                filtered_df['Category'] == category_filter.value
            ]

        # Updated KPI values
        print("TOTAL SALES:", round(filtered_df['Sales'].sum(), 2))
        print("TOTAL PROFIT:", round(filtered_df['Profit'].sum(), 2))
        print("TOTAL ORDERS:", filtered_df['Order ID'].nunique())
        print("AVERAGE SALES:", round(filtered_df['Sales'].mean(), 2))

        # Updated monthly sales chart
        monthly = filtered_df.groupby(
            filtered_df['Order Date'].dt.to_period('M')
        )['Sales'].sum().reset_index()

        monthly['Order Date'] = monthly['Order Date'].astype(str)

        fig1 = px.line(
            monthly,
            x='Order Date',
            y='Sales',
            title='Monthly Sales Trend',
            markers=True
        )
        fig1.show()

        # Updated category chart
        category = filtered_df.groupby('Category')['Sales'].sum().reset_index()

        fig2 = px.bar(
            category,
            x='Category',
            y='Sales',
            color='Category',
            title='Sales by Category'
        )
        fig2.show()

# Connect filters
region_filter.observe(update_dashboard, names='value')
category_filter.observe(update_dashboard, names='value')

# Display dashboard
display(widgets.HBox([region_filter, category_filter]))
display(output)

update_dashboard()
```

```python
# Create date range filters
start_date = widgets.DatePicker(
    description='Start Date:',
    value=df['Order Date'].min().date()
)

end_date = widgets.DatePicker(
    description='End Date:',
    value=df['Order Date'].max().date()
)

def update_dashboard_with_dates(change=None):

    with output:
        clear_output(wait=True)

        filtered_df = df.copy()

        if region_filter.value != 'All':
            filtered_df = filtered_df[
                filtered_df['Region'] == region_filter.value
            ]

        if category_filter.value != 'All':
            filtered_df = filtered_df[
                filtered_df['Category'] == category_filter.value
            ]

        if start_date.value:
            filtered_df = filtered_df[
                filtered_df['Order Date'] >= pd.Timestamp(start_date.value)
            ]

        if end_date.value:
            filtered_df = filtered_df[
                filtered_df['Order Date'] <= (
                    pd.Timestamp(end_date.value) + pd.Timedelta(days=1)
                )
            ]

        print("TOTAL SALES:", round(filtered_df['Sales'].sum(), 2))
        print("TOTAL PROFIT:", round(filtered_df['Profit'].sum(), 2))
        print("TOTAL ORDERS:", filtered_df['Order ID'].nunique())
        print("AVERAGE SALES:", round(filtered_df['Sales'].mean(), 2))

        monthly = filtered_df.groupby(
            filtered_df['Order Date'].dt.to_period('M')
        )['Sales'].sum().reset_index()

        monthly['Order Date'] = monthly['Order Date'].astype(str)

        fig = px.line(
            monthly,
            x='Order Date',
            y='Sales',
            title='Monthly Sales Trend',
            markers=True
        )
        fig.show()

# Connect all filters
region_filter.unobserve(update_dashboard, names='value')
category_filter.unobserve(update_dashboard, names='value')

region_filter.observe(update_dashboard_with_dates, names='value')
category_filter.observe(update_dashboard_with_dates, names='value')
start_date.observe(update_dashboard_with_dates, names='value')
end_date.observe(update_dashboard_with_dates, names='value')

display(widgets.HBox([start_date, end_date]))

update_dashboard_with_dates()
```

```python
# Calculate profit by category
profit_data = df.groupby('Category')['Profit'].sum().reset_index()

# Create interactive pie chart
fig = px.pie(
    profit_data,
    names='Category',
    values='Profit',
    title='Profit Distribution by Category',
    hole=0.4
)

fig.update_traces(textinfo='percent+label')

fig.update_layout(
    template='plotly_white'
)

fig.show()
```

```python
import plotly.express as px
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# Dashboard filters
region = widgets.Dropdown(
    options=['All'] + sorted(df['Region'].unique().tolist()),
    value='All',
    description='Region:'
)

category = widgets.Dropdown(
    options=['All'] + sorted(df['Category'].unique().tolist()),
    value='All',
    description='Category:'
)

start = widgets.DatePicker(
    description='Start:',
    value=df['Order Date'].min().date()
)

end = widgets.DatePicker(
    description='End:',
    value=df['Order Date'].max().date()
)

dashboard_output = widgets.Output()

# Dashboard update function
def create_dashboard(change=None):

    with dashboard_output:
        clear_output(wait=True)

        data = df.copy()

        if region.value != 'All':
            data = data[data['Region'] == region.value]

        if category.value != 'All':
            data = data[data['Category'] == category.value]

        if start.value:
            data = data[data['Order Date'] >= pd.Timestamp(start.value)]

        if end.value:
            data = data[data['Order Date'] <
                        pd.Timestamp(end.value) + pd.Timedelta(days=1)]

        if data.empty:
            display(HTML("<h3>No data available for these filters.</h3>"))
            return

        # KPI calculations
        sales = data['Sales'].sum()
        profit = data['Profit'].sum()
        orders = data['Order ID'].nunique()
        avg_sales = data['Sales'].mean()

        # KPI cards
        cards = f"""
        <div style="display:flex;flex-wrap:wrap;gap:12px;">
        """

        metrics = [
            ("Total Sales", f"${sales:,.2f}", "#1976D2"),
            ("Total Profit", f"${profit:,.2f}", "#269B62"),
            ("Total Orders", f"{orders:,}", "#8055B5"),
            ("Average Sales", f"${avg_sales:,.2f}", "#E58B25")
        ]

        for title, value, color in metrics:
            cards += f"""
            <div style="flex:1;min-width:150px;background:{color};
            color:white;padding:18px;border-radius:10px;">
            <h3>{title}</h3>
            <h2>{value}</h2>
            </div>
            """

        cards += "</div>"
        display(HTML(cards))

        # Monthly sales
        monthly = data.groupby(
            data['Order Date'].dt.to_period('M')
        )['Sales'].sum().reset_index()

        monthly['Order Date'] = monthly['Order Date'].astype(str)

        fig1 = px.line(
            monthly,
            x='Order Date',
            y='Sales',
            title='Monthly Sales Trend',
            markers=True
        )
        fig1.update_layout(template='plotly_white')
        fig1.show()

        # Category sales
        cat = data.groupby('Category')['Sales'].sum().reset_index()

        fig2 = px.bar(
            cat,
            x='Category',
            y='Sales',
            color='Category',
            title='Sales by Category'
        )
        fig2.update_layout(template='plotly_white')
        fig2.show()

        # Regional profit
        reg = data.groupby('Region')['Profit'].sum().reset_index()

        fig3 = px.bar(
            reg,
            x='Region',
            y='Profit',
            color='Region',
            title='Profit by Region'
        )
        fig3.update_layout(template='plotly_white')
        fig3.show()

        # Top 10 products
        top = data.groupby('Product Name')['Sales'].sum().nlargest(10).reset_index()

        fig4 = px.bar(
            top,
            x='Sales',
            y='Product Name',
            orientation='h',
            title='Top 10 Products by Sales'
        )
        fig4.update_layout(
            template='plotly_white',
            yaxis={'categoryorder': 'total ascending'}
        )
        fig4.show()

        # Profit distribution
        profit_cat = data.groupby('Category')['Profit'].sum().reset_index()

        fig5 = px.pie(
            profit_cat,
            names='Category',
            values='Profit',
            hole=0.4,
            title='Profit Distribution'
        )
        fig5.show()


# Connect filters
for control in [region, category, start, end]:
    control.observe(create_dashboard, names='value')

# Dashboard title and filters
display(HTML("""
<h1 style="text-align:center;color:#1976D2;">
SUPERSTORE SALES ANALYTICS DASHBOARD
</h1>
<p style="text-align:center;">
Interactive Business Performance Analysis
</p>
"""))

display(widgets.HBox([region, category]))
display(widgets.HBox([start, end]))
display(dashboard_output)

# Initial dashboard
create_dashboard()
```
